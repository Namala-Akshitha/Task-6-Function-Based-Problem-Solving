```python
def analyze_seat_booking(total_seats, booked_list):
    booked_count = len(booked_list)
    #calculate occupancy percentage
    occupancy = (booked_count / total_seats) * 100
    print("Occupancy:", int(occupancy), "%")
    #Determine show status
    if occupancy == 100:
        print("Show Status: Housefull")
    elif occupancy >= 75:
        print("Show Status: Almost Full")
    elif occupancy >= 50:
        print("Show Status: Filling Fast")
    else:
        print("Show Status: Seats Available")
    #Suggest additional show
    if occupancy >= 90:
        print("Suggestion: Consider Opening Additional Show")
#USER INPUT
total = int(input("Total Seats: "))
booked_number = int(input("Number of Booked Seats: "))
#Create booked seats list
booked = [1] * booked_number
analyze_seat_booking(total, booked)
```
