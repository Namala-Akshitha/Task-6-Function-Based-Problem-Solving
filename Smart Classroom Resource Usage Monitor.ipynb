```python
def track_resource_usage(resource_dict):
    overused = []
    #Check each resource
    for resource in resource_dict:
        if resource_dict[resource] > 8:   # Overuse condition
            overused.append(resource)
    #Output results
    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")
#USER INPUT
resources = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}
track_resource_usage(resources)
```
